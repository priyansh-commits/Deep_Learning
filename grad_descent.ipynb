```python
import numpy as np
import pandas as pd
import time
```

```python
df=pd.read_csv('/content/placement.csv')
df.head(5)
```

```python
X=df.iloc[:,0]
y=df.iloc[:,-1]
```

```python
y
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X.values.reshape(-1, 1))
```

```python
X_scaled=pd.DataFrame(X_scaled)
X_scaled.shape
```

```python
import tensorflow as tf
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model=Sequential()
model.add(Dense(1,activation='linear',input_dim=1))
model.add(Dense(2,activation='relu',input_dim=1))
model.add(Dense(1,activation='linear',input_dim=1))
```

```python
model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='mse',  # Loss function for training
    metrics=['mse', 'mae']  # Metrics for monitoring
)
start = time.time()
history = model.fit(X_scaled,y,epochs=10,batch_size=1,validation_split=0.2)
print(time.time() - start)
```

```python
y_pred=model.predict(y)
from sklearn import metrics
print('MAE:', metrics.mean_absolute_error(y, y_pred))
print('MSE:', metrics.mean_squared_error(y, y_pred))
```
